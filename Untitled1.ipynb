# A

The distinction between "contextual" and "functional" words in the papers is contextual words change significantly depending on the subject matter. While functional are words that have little meaning on their own. They're words like "the" "of" "by". Mosteller and Wallace chose to analyze function words. They chose them since it reads better into the author's habits and work like their author signature and tendencies.

# B

```python
fword_count = pd.read_csv("shared/federalist/function_word_counts.csv")
fword_count
```

```python
fword_count["all"].value_counts()
```

```python
hamilton = fword_count[(fword_count["Author"] == "Hamilton")][["all", "total_word_count"]]
hamilton
```

```python
madison = fword_count[fword_count["Author"] == "Madison"][["all", "total_word_count"]]
madison
```

```python
madison["madcountword"] = madison["all"] / madison["total_word_count"]
madrelative_frequency = madison.sort_values(by="madcountword", ascending=True)
len(madrelative_frequency)
madrelative_frequency
```

```python
hamilton["hamcountword"] = hamilton["all"] / hamilton["total_word_count"]
hamrelative_frequency = hamilton.sort_values(by="hamcountword", ascending=True)
len(hamrelative_frequency)
hamrelative_frequency
```

# The relative frequency of the word "all" in Hamilton and Madison texts is about .4% and madrelative_frequency and hamrelative_frequency order each essays relative frequency of the functional word all in order from least to most.

# C

```python
binmin = min(madison["madcountword"].min(), hamilton["hamcountword"].min())
binmin
binmax = max(hamilton["hamcountword"].max(), madison["madcountword"].max())
binmax

bin_width = 0.0005
shared_bin = np.arange(binmin, binmax + bin_width, bin_width)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(3, 7), sharex=True)


ax[0].hist(hamilton["hamcountword"], bins=shared_bin)
ax[0].set_title("hamilton essays")

ax[1].hist(madison["madcountword"], bins=shared_bin)
ax[1].set_title("madison essays")
ax[1].set_xlabel("all relative freq")
```
